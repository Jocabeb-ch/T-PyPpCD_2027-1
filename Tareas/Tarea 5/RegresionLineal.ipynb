## Código de clase

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Datos : tamano en m^2 (X) y precio en miles (y)
X = np.array([50, 60, 70, 80, 90, 100, 110, 120])
y = np.array([150, 180, 195, 230, 260, 290, 325, 355])

m = X.size
X_b = np.c_[np.ones((m,1)),X] # columna de unos ( theta_0 )
```

```python
# Ecuacion normal : theta = (X^T X)^ -1 X^T y
theta = np.linalg.inv(X_b.T @ X_b) @X_b.T @ y
print("Pendiente :", round (theta [1],2))
print("Intercepto :", round (theta [0],2))
```

```python
# Prediccion para una casa de 95 m^2
precio=theta[0] + theta[1] * 95
print ("Precio estimado:", round(precio,2), "miles")
```

##Scatter Plot

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(
    X,
    y,
    color = "purple"
)

x_g = np.arange(X.min(), X.max(), step = 0.5)
y_g = theta[0] + theta[1] * x_g

ax.plot(
    x_g,
    y_g,
    color = "#fe46a5",
    linestyle = "--",
    linewidth = 2,
)
ax.set_xlabel("Tamaño en Metros Cuadrados")
ax.set_ylabel("Precio en miles")
ax.set_title("Estimación de precio de vivienda por metros cuadrados")
plt.show()
```
